## Read data from .csv files to separate data frames 

```python
# Define path to data
path = 'Data/'          #Das ist einfach der Dateipfad zu den CSV-Dateien (siehe links im Folder), 
                        #der nun der variable path zugewiesen wird. Case-sensitive!

# Read data from .csv files
#Gemäss dem Format zum CSV einlesen data = pd.read_csv("datei.csv", sep=",", encoding="utf-8")
df_customers = pd.read_csv(f'{path}Customers.csv', sep=',')
df_categories = pd.read_csv(f'{path}Categories.csv', sep=',')
df_employees = pd.read_csv(f'{path}Employees.csv', sep=',')
df_order_details = pd.read_csv(f'{path}OrderDetails.csv', sep = ',')
df_orders = pd.read_csv(f'{path}Orders.csv', sep=',')
df_products = pd.read_csv(f'{path}Products.csv', sep=',')
df_shippers = pd.read_csv(f'{path}Shippers.csv', sep=',')
df_suppliers = pd.read_csv(f'{path}Suppliers.csv', sep=',')

# Change column names to lowercase
#Format: dataframe.columns = dataframe.columns.str.lower()
df_customers.columns = df_customers.columns.str.lower()
df_categories.columns = df_categories.columns.str.lower()
df_employees.columns = df_employees.columns.str.lower()
df_order_details.columns = df_order_details.columns.str.lower()
df_orders.columns = df_orders.columns.str.lower()
df_products.columns = df_products.columns.str.lower()
df_shippers.columns = df_shippers.columns.str.lower()
df_suppliers.columns = df_suppliers.columns.str.lower()

# Check dimensions of DataFrames
#Grösse (Anzahl Zeilen & Spalten) der DataFrames anzeigen
#Format: dataframe.shape
#Ohne F-String: print("Customers:", df_customers.shape)
#Mit F-String: print(f"Customers: {df_customers.shape}")
#F-String baut das zurückgegebene Tupel direkt in den Text ein
print("Dimensions of DataFrames:")
print(f'Customers: {df_customers.shape}')
print(f'Categories: {df_categories.shape}')
print(f'Employees: {df_employees.shape}')
print(f'Order Details: {df_order_details.shape}')
print(f'Orders: {df_orders.shape}')
print(f'Products: {df_products.shape}')
print(f'Shippers: {df_shippers.shape}')
print(f'Suppliers: {df_suppliers.shape}')
```

```python
# Set up database connection
#Direkt aus dem Docker-Compose-File holen
user = "pgadmin"
password = "geheim"
host = "localhost"
port = "5432"
database = "northwind"

# Create Connection URL
db_connection_url = "postgresql://" + user + ":" + password +\
                    "@" + host + ":" + port + "/" + database

# Create SQLAlchemy Engine
engine = create_engine(db_connection_url)

# Test database connection
with engine.connect() as connection:
    result = connection.execute(text('SELECT current_database()'))
    print(result.fetchone())

# Dispose the engine
engine.dispose()
```

## Write data frames to database tables

```python
# Create SQLAlchemy Engine
#Format engine = create_engine(db_connection_url)
#URL haben wir ja bereits: 
#db_connection_url = "postgresql://" + user + ":" + password + \
#"@" + host + ":" + port + "/" + database
engine = create_engine(db_connection_url)   #Erstellt ein SQLAlchemy-Engine-Objekt für die Verbindung zur definierten Datenbank

# Drop tables if it exist
#Falls die Tabellen schon in der Datenbank existieren, sollen sie zuerst gelöscht werden
#Grund: Danach werden die Tabellen aus deinen DataFrames neu erstellt; ohne Löschen könnte es Konflikte oder doppelte Daten geben
    #engine.connect() → öffnet die Datenbankverbindung
    #as connection → gibt dieser Verbindung den Namen connection
    #with → sorgt dafür, dass die Verbindung nach dem eingerückten Block automatisch wieder geschlossen wird
with engine.connect() as connection:
    connection.execute(text("DROP TABLE IF EXISTS public.Categories CASCADE;"))
    connection.execute(text("DROP TABLE IF EXISTS public.Customers CASCADE;"))
    connection.execute(text("DROP TABLE IF EXISTS public.Employees CASCADE;"))
    connection.execute(text("DROP TABLE IF EXISTS public.OrderDetails CASCADE;"))
    connection.execute(text("DROP TABLE IF EXISTS public.Orders CASCADE;"))
    connection.execute(text("DROP TABLE IF EXISTS public.Products CASCADE;"))
    connection.execute(text("DROP TABLE IF EXISTS public.Shippers CASCADE;"))
    connection.execute(text("DROP TABLE IF EXISTS public.Suppliers CASCADE;"))
    connection.commit() #bestätigt die ausgeführten Änderungen dauerhaft in der Datenbank
    #connection → aktive Datenbankverbindung
    #.execute(...) → führt einen SQL-Befehl aus
    #text(...) → macht aus dem SQL-String ein SQLAlchemy-SQL-Objekt
    #"DROP TABLE IF EXISTS ..." → der eigentliche SQL-Befehl
    #public.Categories → Tabelle Categories im Schema public
    #CASCADE → löscht auch abhängige Beziehungen/Objekte

# Write DataFrames to database
#Speichert die pandas-DataFrames, die aktuell nur in Python existieren, als echte Tabellen in PostgreSQL
#Format dataframe.to_sql(name="tabellenname", con=engine)
df_categories.to_sql('categories', engine)
df_customers.to_sql('customers', engine)
df_employees.to_sql('employees', engine)
df_order_details.to_sql('orderdetails', engine)
df_orders.to_sql('orders', engine)
df_products.to_sql('products', engine)
df_shippers.to_sql('shippers', engine)
df_suppliers.to_sql('suppliers', engine)

# Dispose the engine
#Engine schliessen, Ressourcen freigeben
#Format <engine_variable>.dispose()
engine.dispose()
```

## List tables in the database

```python
# Create SQLAlchemy Engine
engine = create_engine(db_connection_url)

# Open a connection
with engine.connect() as connection:

    # Execute the query
    result = connection.execute(text("""SELECT table_name
                                        FROM information_schema.tables
                                        WHERE table_schema = 'public'"""))
    
    # Fetch and print the results
    for row in result:
        print(row[0])

# Dispose the engine
engine.dispose()
```

## Create a SQL query to retrieve product information from a database.

```python
# Create SQLAlchemy Engine
engine = create_engine(db_connection_url)

# Write data to table
#pd.read_sql_query() → führt eine SQL-Abfrage aus und lädt das Ergebnis als DataFrame
#''' ... ''' → hier kommt deine SQL-Query hinein
#con=engine → Query wird über deine SQLAlchemy-Engine auf PostgreSQL ausgeführt
#df_sub → speichert das Ergebnis als DataFrame

#Die eigentliche Aufgabe steckt nur in der SQL-Query
#Ich brauche infos aus den Tabellen products (Produkt-ID, Produktname, Preis, CategoryID)
#& categories (Kategoriename). Verbunden werden sie über categoryid

df_sub = pd.read_sql_query(     #Die drei Anführungszeichen ''' ... ''' erzeugen in Python einen mehrzeiligen String
        '''SELECT productid, productname, price, products.categoryid, categoryname
        FROM products
        INNER JOIN categories
        ON products.categoryid = categories.categoryid;''',
con=engine)

# Dispose the engine
engine.dispose()    #schliesst die SQLAlchemy-Engine und gibt die Datenbankverbindungen frei

# Show the data
df_sub              #zeigt im Jupyter Notebook den Inhalt des DataFrames df_sub an
```

## Plot product prices

```python
# Plot Histogram
fig = plt.figure( figsize=(7,4))
plt.xticks(fontsize=14, rotation=0)
plt.yticks(fontsize=14, rotation=0)
n, bins, patches = plt.hist(x=df_sub['price'], 
                            bins=20, 
                            color='#C53115',
                            alpha=1.00, 
                            rwidth=0.95
                   )
plt.grid(True)
plt.ticklabel_format(style='plain')
plt.grid(axis='y', alpha=0.75)

# Set labels
plt.xlabel('Price (USD)', fontsize=10, labelpad=10)
plt.ylabel('Frequency', fontsize=10, labelpad=10)
plt.title('Histogram of product prices', fontsize=12, pad=10)

# Set fontsize of tick labels
plt.xticks(fontsize = 10)
plt.yticks(fontsize = 10)

plt.show()
```
